```python
# Week 1 - Day 5
## Feature Engineering
```

```python
# Convert date column to datetime
calendar["date"] = pd.to_datetime(calendar["date"], errors="coerce")

# Check the datatype
print(calendar["date"].dtype)
```

```python
calendar["Year"] = calendar["date"].dt.year
calendar["Month"] = calendar["date"].dt.month
calendar["Day"] = calendar["date"].dt.day
calendar["Weekday"] = calendar["date"].dt.day_name()
calendar["Quarter"] = calendar["date"].dt.quarter
```

```python
calendar[["date", "Year", "Month", "Day", "Weekday", "Quarter"]].head()
```

```python
## Basic Data Analysis
```

```python
daily_sales = sales.iloc[:, 6:].sum()

daily_sales.head(30)
```

```python
plt.figure(figsize=(12,5))
daily_sales.head(30).plot()

plt.title("Total Sales - First 30 Days")
plt.xlabel("Day")
plt.ylabel("Total Sales")

plt.show()
```

```python
calendar["Month"].value_counts().sort_index()
```

```python
calendar["Month"].value_counts().sort_index().plot(kind="bar")

plt.title("Records by Month")
plt.xlabel("Month")
plt.ylabel("Count")

plt.show()
```
